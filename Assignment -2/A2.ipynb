22. Store the monthly electricity bill of six months in a dictionary. Find the month with the highest bill.

```python
months = ["January", "February", "March", "April", "May", "June"]

bills = {}

for month in months:
    bill = int(input(f"Enter electricity bill for {month}: "))
    bills[month] = bill

highest_month = max(bills, key=bills.get)

print("\nHighest Electricity Bill")
print("Month :", highest_month)
print("Bill  : ₹" + str(bills[highest_month]))
```

23. Write a Python program to accept two sets of subjects chosen by two students and find: Common subjects, Subjects only chosen by the first student, Union of all chosen subjects.

```python
n1 = int(input("Enter number of subjects chosen by Student 1: "))
student1 = set()
for i in range(n1):
    subject = input(f"Enter Subject {i+1}: ")
    student1.add(subject)

n2 = int(input("Enter number of subjects chosen by Student 2: "))
student2 = set()
for i in range(n2):
    subject = input(f"Enter Subject {i+1}: ")
    student2.add(subject)

print("\nStudent 1 Subjects:")
print(student1)

print("Student 2 Subjects:")
print(student2)

print("\nCommon Subjects:")
print(student1 & student2)

print("\nSubjects only chosen by Student 1:")
print(student1 - student2)

print("\nUnion of all Subjects:")
print(student1 | student2)
```

24. Given a set of all roll numbers and another set of students present in class, write a Python program to find and display the roll numbers of absent students.

```python
n = int(input("Enter total number of students: "))

print("\nEnter all roll numbers:")
all_roll = set()
for i in range(n):
    roll = int(input())
    all_roll.add(roll)

p = int(input("\nEnter number of students present: "))

print("\nEnter present roll numbers:")
present = set()
for i in range(p):
    roll = int(input())
    present.add(roll)

absent = all_roll - present

print("\nAll Roll Numbers:")
print(all_roll)

print("\nPresent Students:")
print(present)

print("\nAbsent Students:")
print(absent)
```

30. Write a Python function multiply_all(*args) using arbitrary positional arguments (*args) to multiply any number of given integers.

```python
def multiply_all(*args):
    result = 1
    for num in args:
        result *= num
    return result

print("Output 1 =", multiply_all(2, 3, 4))
print("Output 2 =", multiply_all(5, 10))
```

31. Write a Python menu-driven program using user-defined functions for basic calculator operations (Add, Subtract, Multiply, Divide) that executes continuously until the user chooses to exit.

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b

while True:
    print("\n1. Add\n2. Subtract\n3. Multiply\n4. Divide\n5. Exit")
    choice = int(input("Select Choice: "))

    if choice == 5:
        break

    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))

    if choice == 1:
        print("Result:", a, "+", b, "=", add(a, b))
    elif choice == 2:
        print("Result:", a, "-", b, "=", subtract(a, b))
    elif choice == 3:
        print("Result:", a, "*", b, "=", multiply(a, b))
    elif choice == 4:
        print("Result:", a, "/", b, "=", divide(a, b))
    else:
        print("Invalid Choice")
```

32. Write a Python function calculate_grade(marks_list) that accepts a list of subject marks, calculates the percentage, and returns the Grade (A/B/C/F) along with pass/fail status.

```python
def calculate_grade(marks_list):
    percentage = sum(marks_list) / len(marks_list)

    if percentage >= 75:
        grade = "A"
    elif percentage >= 60:
        grade = "B"
    elif percentage >= 40:
        grade = "C"
    else:
        grade = "F"

    status = "PASSED" if percentage >= 40 else "FAILED"

    return percentage, grade, status

percentage, grade, status = calculate_grade([80, 75, 90, 85, 70])
print(f"Percentage: {percentage}% | Grade: {grade} | Status: {status}")
```
