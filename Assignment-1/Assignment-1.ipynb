### 25. Print Fibonacci series using while loop

```python
n = int(input("Enter number of terms: "))
a = 0
b = 1
count = 0
while count < n:
    print(a, end=" ")
    c = a + b
    a = b
    b = c
    count += 1
```

### 26. Check if a number is Prime or Not Prime using for loop

```python
n = int(input("Enter a number: "))
if n > 1:
    prime = True
    for i in range(2, n):
        if n % i == 0:
            prime = False
            break
    if prime:
        print("Prime")
    else:
        print("Not Prime")
else:
    print("Not Prime")
```

### 27. Check whether a year is a Leap Year or Non-Leap Year

```python
year = int(input("Enter a year: "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("Leap Year")
else:
    print("Non-Leap Year")
```
